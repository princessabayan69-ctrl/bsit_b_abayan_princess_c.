```python

# 1. Sum the array
def sum_array(numbers):
    if len(numbers) == 0:
        return 0

    return numbers[0] + sum_array(numbers[1:])


# 2. Find the largest number
def largest_number(numbers):
    if len(numbers) == 1:
        return numbers[0]

    return max(numbers[0], largest_number(numbers[1:]))


# 3. Count a chosen number
def count_number(numbers, target):
    if len(numbers) == 0:
        return 0

    if numbers[0] == target:
        return 1 + count_number(numbers[1:], target)

    return 0 + count_number(numbers[1:], target)


# Testing the functions

print("1. Sum of Array")
print(sum_array([]))
print(sum_array([5]))
print(sum_array([2, 4, 6]))
print(sum_array([10, -5, 3]))

print("\n2. Largest Number")
print(largest_number([5]))
print(largest_number([20, 10, 5]))
print(largest_number([3, 8, 15]))
print(largest_number([-5, -2, -10]))

print("\n3. Count a Number")
print(count_number([], 2))
print(count_number([2, 5, 2, 7, 2], 2))
print(count_number([1, 3, 4, 5], 2))
```
